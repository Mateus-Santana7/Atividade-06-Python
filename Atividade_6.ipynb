```python
#atividade02
def soma_lista(numeros):
  soma = 0
  for numero in numeros:
    soma += numero
  return soma

minha_lista = [10, 20, 30, 40]
resultado_soma = soma_lista(minha_lista)
print(f"A soma dos elementos da lista {minha_lista} é: {resultado_soma}")
```

```python
## atividade 02
def soma_lista(numeros):
  soma = 0
  for numero in numeros:
    soma = soma + numero
  return soma
numeros = [10, 20, 30, 40]
print(soma_lista(numeros))
```

```python
##atividade 03
def filtrar_maiores_de_idade(idades):
  maiores = []
  for idade in idades:
    if idade >= 18:
      maiores.append(idade)
  return maiores
idades = [15,18,20,16,25]
print(filtrar_maiores_de_idade(idades))
```

```python
#atividade 04
def calcular_juros(valor, taxa=0.05):
  juros = valor * taxa
  return valor + juros
print(calcular_juros(1000))
print(calcular_juros(1000,0.10))
```

```python
## atividade 05
def analisar_notas(notas):

    maior = notas[0]
    menor = notas[0]
    soma = 0
    aprovados = 0

    for nota in notas:

        if nota > maior:
            maior = nota

        if nota < menor:
            menor = nota

        soma += nota

        if nota >= 6:
            aprovados += 1

    media = soma / len(notas)

    resultado = {
        "Maior nota": maior,
        "Menor nota": menor,
        "Média": media,
        "Aprovados": aprovados
    }

    return resultado


notas = [7, 8, 5, 9, 4, 6]

resultado = analisar_notas(notas)

print(resultado)
```
